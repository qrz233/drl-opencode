### Channel model

```python
n_points = 100
zmax, zmin = z_max, z_min
xuav, yuav = 0, 0
xaxis = np.linspace(0, boundary, n_points)
yaxis = np.zeros(n_points)
alpha = 1

print(f"bw per user = {bw_per_user / 1e6:.1f} MHz")
```

```python
def cal_capacity_zuav(xaxis, yaxis, xuav, yuav, zuav, n_points=n_points):
    '''Calculate the downlink throughput w.r.t. the relative position of users'''
    chgain = np.zeros(n_points)
    chcapa_Mb = np.zeros(n_points)
    chcapa_bpsHz = np.zeros(n_points)
    for i in range(0,n_points):
        chgain[i] = cal_channel_fading_gain(xaxis[i], yaxis[i], xuav, yuav, zuav)
        chcapa_Mb[i], chcapa_bpsHz[i] = cal_channel_capacity_Mb(chgain[i], alpha)
    return chcapa_Mb, chcapa_bpsHz
```

```python
def cal_capacity_xyuser(xuser, yuser, zuav_vec):
    chgain = np.zeros(n_points)
    chcapa_Mb = np.zeros(n_points)
    chcapa_bpsHz = np.zeros(n_points)
    for i in range(0,n_points):
        chgain[i] = cal_channel_fading_gain(xuser, yuser, xuav, yuav, zuav_vec[i])
        chcapa_Mb[i], chcapa_bpsHz[i] = cal_channel_capacity_Mb(chgain[i], alpha)
    
    return chcapa_Mb, chcapa_bpsHz
```

### Stationary

```python
# Initiate the UAV Swarm 
uavswarm_stationary = [UAV(name= f'UAV {i} (Stationary)', x_init=init_locations_uav[i][0], y_init=init_locations_uav[i][1], z_init=z_init, 
                xlim = xlim_uav[i], ylim = ylim_uav[i], speedxy_max=0, speedz_max=0) for i in range(0,n_uavs)]

# Initiate users of the Stationary UAV Swarm with movements and traffic 
users_stationary = [User(f'User {i}') for i in range(n_users)]
load_user_properties(users_stationary, xaxis_all, yaxis_all, arrival_traffic_Mb, traffic_state, traffic_type, ema_traffic_Mb, lambd_Mb_alluser)

# Alias for ease of code maintenance 
uavs = uavswarm_stationary
users = users_stationary          
    
# Simulation starts!  
bh_capa_log, bh_scale_log = [], []
for t in range(0, n_slots):         # t = timeslot
    if t%100==0: 
        print(f"t={t}")
    # (0) Collect system statistics 
    queue_length_Mb = np.array([user.queue_length_Mb[t-1] for user in users]) if t>0 else np.zeros(n_users)    # shape=(n_users,)
    traffic_Mb = np.array([user.arrival_traffic_Mb[t] for user in users])       # shape=(n_users,)
    isActive_all = find_active_users(queue_len_Mb=queue_length_Mb, incoming_traffic_Mb=traffic_Mb)  # shape=(n_users,), =1 if qlen + traffic > 0 
    mos_all = np.array([user.mos[t-1] for user in users]) if t > 0 else 5 * np.ones(n_users)

    # (1) User Clustering 
    loc_users = np.hstack((np.expand_dims(xaxis_all[:,t], axis=1),
                           np.expand_dims(yaxis_all[:,t], axis=1),
                           np.zeros(shape=(n_users,1))))                    # shape = (n_users, 3)
    loc_uavs = np.array([[uav.x[t-1], uav.y[t-1], uav.z[t-1]] for uav in uavs]) if t>0 \
                else  np.array([[uav.x[0], uav.y[0], uav.z[0]] for uav in uavs])  # shape = (n_uavs, 3)
    cmat, cids, c_centroids = kmeans_clustering(loc_users)        # cluster matrix, ids, and centroid
    
    if t==0:  # clustering at t=0 
        cids0 = cids.copy()
    
    # # (2) Bandwidth allocation based on the traffic demand 
    # alpha_uavs = np.zeros(shape=(n_uavs, n_users))      # shape=(n_uavs, n_users)
    # for cid, uav in enumerate(uavs):   
    #     mask = isActive_all & (cids==cid)               # users that are active and belong to the considered cluster, shape=(n_users,) 
    #     alpha_uavs[cid] = allocate_bandwidth_based_on_demand(queue_length_Mb+traffic_Mb, mask)
    # assert np.all(np.sum(alpha_uavs>0, axis=0) <= 1), "Each user should be associated to exactly one UAV-BS"
    # bw_alloc_all = np.sum(alpha_uavs, axis=0)              # shape=(n_users,)
    
    # # (2) Bandwidth Allocation: allocate equally among active users (qlen + new traffic > 0) 
    # bw_alloc_all = allocate_bandwidth_equal_to_active_users(isActive_all)
    
    # (2) Bandwidth Allocation (modified): the 50 MHz pool is split across the 3 service
    # types with shares 30% (Web) / 60% (Video) / 10% (Messaging); in each slot a type's
    # pool is shared equally by its active users. alpha = user bw / bw_per_user (reference),
    # so channel capacity = alpha * bw_per_user * log2(1+SNR) = user_bw * log2(1+SNR).
    bw_alloc_Hz_all = np.zeros(shape=(n_users,))
    ttype_all_t = traffic_type[:, t]
    for k in range(len(bw_type_pool_Hz)): 
        ids_k = np.where(ttype_all_t == k)[0]
        active_k = ids_k[isActive_all[ids_k]]
        if len(active_k) > 0: 
            bw_alloc_Hz_all[active_k] = bw_type_pool_Hz[k] / len(active_k)
    bw_alloc_all = bw_alloc_Hz_all / bw_per_user      # fraction of the reference per-user bw
    
    # (3) UAV Updates: Keep stationary at the first location 
    # for uav in uavs: 
    #     uav.x[t], uav.y[t], uav.z[t] = uav.x[0], uav.y[0], uav.z[0] 
        
    # (3) UAV Updates: Move toward the initial centroids (of t=0) 
    if t > 0 and (not uav_fixed_placement): 
        for cid, uav in enumerate(uavs):        # cid = cluster id 
            if t < t_training_start: 
                dx, dy, dd, uav_speed = 0, 0, 0, 0  # no movement 
            else: 
                xc0, yc0 = np.mean(xaxis_all[cids0==cid, 0]), np.mean(yaxis_all[cids0==cid, 0])   
                dx, dy, dd = cal_direction(uav.x[t-1], uav.y[t-1], xc0, yc0)
                # uav_speed = np.min([dd/(uav.speedxy_max*slot_len), 1])*uav.speedxy_max      # greedily move toward the centroid w/ vmax 
                uav_speed = np.min([dd/(uav_vxy_fixed*slot_len), 1])*uav_vxy_fixed
            
            uav.x[t], uav.y[t] = update_UAV_location(
                x0=uav.x[t-1], y0=uav.y[t-1], dx=dx, dy=dy, speed=uav_speed, slot_len=slot_len
            )
            uav.z[t] = uav.z[0]
            uav.velocity[t] = uav_speed
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=uav_speed)[0]
        
    # Modified: aerial BSs are pinned at the center of the area (altitude fixed at 300 m)
    if uav_fixed_placement: 
        for uav in uavs: 
            uav.x[t], uav.y[t], uav.z[t] = uav.x[0], uav.y[0], uav.z[0]
            uav.velocity[t] = 0
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=0)[0]
    # (3.5) Modified: backhaul constraint - airship -> drone k hard-caps the drone's
    # total served downlink rate (excess stays queued; bandwidth allocation untouched).
    bh_scale = np.ones(shape=(n_uavs,))
    bh_capa_k = np.zeros(shape=(n_uavs,))
    for k in range(n_uavs): 
        ids_k = np.where(cids == k)[0]
        if len(ids_k) == 0: 
            continue
        g_k = cal_channel_fading_gain(xaxis_all[ids_k, t], yaxis_all[ids_k, t], uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        capa_k = bw_alloc_Hz_all[ids_k] * np.log2(1 + pTx_downlink * g_k / noise_pw_total) * slot_len / 1e6   # Mb/slot
        offered_k = np.sum(np.minimum(capa_k, queue_length_Mb[ids_k] + traffic_Mb[ids_k]))
        bh_capa_k[k] = cal_backhaul_capacity_Mb(uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        if offered_k > bh_capa_k[k]: 
            bh_scale[k] = bh_capa_k[k] / offered_k
    bh_capa_log.append(np.mean(bh_capa_k))
    bh_scale_log.append(np.min(bh_scale))
    # (4) User Updates  
    for id, user in enumerate(users): 
        user.cluster[t] = cids[id]; cid = cids[id]      # id of the user's cluster  
        user.active_state[t] = isActive_all[id]
        user.alpha[t] = bw_alloc_all[id]               
        user.chgains_downlink[t] = cal_channel_fading_gain(user.x[t], user.y[t], uavs[cid].x[t], uavs[cid].y[t], uavs[cid].z[t])
        user.channel_capacity_Mb[t] = cal_channel_capacity_Mb(user.chgains_downlink[t], user.alpha[t])[0]
        
        qlen_prev_Mb = user.queue_length_Mb[t-1] if t > 0 else 0
        depart_Mb = cal_downlink_throughput_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], user.channel_capacity_Mb[t]) * bh_scale[cid]
        user.queue_length_Mb[t] = update_queue_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], depart_Mb)
        user.downlink_throughput_Mb[t] = depart_Mb
        user.ema_qlen_Mb[t] = alpha_ema*user.queue_length_Mb[t] + (1-alpha_ema)*user.ema_qlen_Mb[t-1]
        ttype_t = int(user.traffic_type[t])   # service type active in this slot (-1 if OFF)
        ld_type_Mbps = ON_data_arrival_mean_Mb[ttype_t] if ttype_t >= 0 else user.lambd_Mb
        user.mos[t] = mos_func(qlen=user.ema_qlen_Mb[t], lambda_=ld_type_Mbps, ss_min_arr=ss_min_types[ttype_t], ss_max_arr=ss_max_types[ttype_t]) if (user.traffic_state[t]==True) else np.nan
        user.delay[t] = d0 + user.ema_qlen_Mb[t] / ld_type_Mbps

print('Backhaul [Stationary]: avg capacity/drone = ' +
      f'{np.mean(bh_capa_log):.1f} Mb/slot, time-avg of min bh_scale = {np.mean(bh_scale_log):.3f} (1.0 = never throttled)')
```

### Centroid

```python
# Initiate the UAV Swarm 
uav_swarm_centroid = [UAV(name= f'UAV {i} (Centroid)', x_init=init_locations_uav[i][0], y_init=init_locations_uav[i][1], z_init=z_init, 
                        xlim = xlim_uav[i], ylim = ylim_uav[i], speedxy_max=uav_speedxy_max, speedz_max=uav_speedz_max) for i in range(0,n_uavs)]

# Initiate users of the Stationary UAV Swarm with movements and traffic 
users_centroid = [User(f'User {i}') for i in range(n_users)]
load_user_properties(users_centroid, xaxis_all, yaxis_all, arrival_traffic_Mb, traffic_state, traffic_type, ema_traffic_Mb, lambd_Mb_alluser)

# Alias for ease of code maintenance 
uavs = uav_swarm_centroid
users = users_centroid          
    
# Simulation starts!  
bh_capa_log, bh_scale_log = [], []
for t in range(0, n_slots):         # t = timeslot
    if t%100==0: 
        print(f"t={t}")
    # (0) Collect system statistics 
    queue_length_Mb = np.array([user.queue_length_Mb[t-1] for user in users]) if t>0 else np.zeros(n_users)     # shape=(n_users,)
    traffic_Mb = np.array([user.arrival_traffic_Mb[t] for user in users])       # shape=(n_users,)
    isActive_all = find_active_users(queue_len_Mb=queue_length_Mb, incoming_traffic_Mb=traffic_Mb)  # shape=(n_users,), =1 if qlen + traffic > 0
    mos_all = np.array([user.mos[t-1] for user in users]) if t > 0 else 5 * np.ones(n_users)

    # (1) User Clustering 
    loc_users = np.hstack((np.expand_dims(xaxis_all[:,t], axis=1), np.expand_dims(yaxis_all[:,t], axis=1), np.zeros(shape=(n_users,1))))    # shape = (n_users, 3)
    loc_uavs = np.array([[uav.x[t-1], uav.y[t-1], uav.z[t-1]] for uav in uavs]) if t>0 \
                else  np.array([[uav.x[0], uav.y[0], uav.z[0]] for uav in uavs])  # shape = (n_uavs, 3)
    cmat, cids, c_centroids = kmeans_clustering(loc_users)        # cluster matrix, ids, and centroid
    
    # (2) Bandwidth Allocation (modified): the 50 MHz pool is split across the 3 service
    # types with shares 30% (Web) / 60% (Video) / 10% (Messaging); in each slot a type's
    # pool is shared equally by its active users. alpha = user bw / bw_per_user (reference),
    # so channel capacity = alpha * bw_per_user * log2(1+SNR) = user_bw * log2(1+SNR).
    bw_alloc_Hz_all = np.zeros(shape=(n_users,))
    ttype_all_t = traffic_type[:, t]
    for k in range(len(bw_type_pool_Hz)): 
        ids_k = np.where(ttype_all_t == k)[0]
        active_k = ids_k[isActive_all[ids_k]]
        if len(active_k) > 0: 
            bw_alloc_Hz_all[active_k] = bw_type_pool_Hz[k] / len(active_k)
    bw_alloc_all = bw_alloc_Hz_all / bw_per_user      # fraction of the reference per-user bw
    
    # (3) UAV Updates: Movement and Energy Consumption  
    if t > 0 and (not uav_fixed_placement): 
        for cid, uav in enumerate(uavs):        # cid = cluster id 
            if t < t_training_start: 
                dx, dy, dd, uav_speed = 0, 0, 0, 0  # no movement 
            else: 
                xc, yc = np.mean(xaxis_all[cids==cid, t]), np.mean(yaxis_all[cids==cid, t])   # centroid of the cluster 
                dx, dy, dd = cal_direction(uav.x[t-1], uav.y[t-1], xc, yc)
                uav_speed = np.min([dd/(uav_vxy_fixed * slot_len), 1]) * uav_vxy_fixed 

            uav.x[t], uav.y[t] = update_UAV_location(x0=uav.x[t-1], y0=uav.y[t-1], dx=dx, dy=dy, speed=uav_speed, slot_len=slot_len)
            uav.z[t] = uav.z[0]
            uav.velocity[t] = uav_speed
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=uav_speed)[0]


    # Modified: aerial BSs are pinned at the center of the area (altitude fixed at 300 m)
    if uav_fixed_placement: 
        for uav in uavs: 
            uav.x[t], uav.y[t], uav.z[t] = uav.x[0], uav.y[0], uav.z[0]
            uav.velocity[t] = 0
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=0)[0]
    # (3.5) Modified: backhaul constraint - airship -> drone k hard-caps the drone's
    # total served downlink rate (excess stays queued; bandwidth allocation untouched).
    bh_scale = np.ones(shape=(n_uavs,))
    bh_capa_k = np.zeros(shape=(n_uavs,))
    for k in range(n_uavs): 
        ids_k = np.where(cids == k)[0]
        if len(ids_k) == 0: 
            continue
        g_k = cal_channel_fading_gain(xaxis_all[ids_k, t], yaxis_all[ids_k, t], uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        capa_k = bw_alloc_Hz_all[ids_k] * np.log2(1 + pTx_downlink * g_k / noise_pw_total) * slot_len / 1e6   # Mb/slot
        offered_k = np.sum(np.minimum(capa_k, queue_length_Mb[ids_k] + traffic_Mb[ids_k]))
        bh_capa_k[k] = cal_backhaul_capacity_Mb(uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        if offered_k > bh_capa_k[k]: 
            bh_scale[k] = bh_capa_k[k] / offered_k
    bh_capa_log.append(np.mean(bh_capa_k))
    bh_scale_log.append(np.min(bh_scale))
    # (4) User Updates  
    for id, user in enumerate(users): 
        user.cluster[t] = cids[id]
        cid = cids[id]      # id of the user's cluster  
        user.active_state[t] = isActive_all[id]
        user.alpha[t] = bw_alloc_all[id]       # fixed bandwidth for each user 
        user.chgains_downlink[t] = cal_channel_fading_gain(user.x[t], user.y[t], uavs[cid].x[t], uavs[cid].y[t], uavs[cid].z[t])
        user.channel_capacity_Mb[t] = cal_channel_capacity_Mb(user.chgains_downlink[t], user.alpha[t])[0]
        
        qlen_prev_Mb = user.queue_length_Mb[t-1] if t > 0 else 0
        depart_Mb = cal_downlink_throughput_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], user.channel_capacity_Mb[t]) * bh_scale[cid]
        user.queue_length_Mb[t] = update_queue_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], depart_Mb)
        user.downlink_throughput_Mb[t] = depart_Mb
        user.ema_qlen_Mb[t] = alpha_ema*user.queue_length_Mb[t] + (1-alpha_ema)*user.ema_qlen_Mb[t-1]
        ttype_t = int(user.traffic_type[t])   # service type active in this slot (-1 if OFF)
        ld_type_Mbps = ON_data_arrival_mean_Mb[ttype_t] if ttype_t >= 0 else user.lambd_Mb
        user.mos[t] = mos_func(qlen=user.ema_qlen_Mb[t], lambda_=ld_type_Mbps, ss_min_arr=ss_min_types[ttype_t], ss_max_arr=ss_max_types[ttype_t]) if (user.traffic_state[t]==True) else np.nan
        user.delay[t] = d0 + user.ema_qlen_Mb[t] / ld_type_Mbps

print('Backhaul [Centroid]: avg capacity/drone = ' +
      f'{np.mean(bh_capa_log):.1f} Mb/slot, time-avg of min bh_scale = {np.mean(bh_scale_log):.3f} (1.0 = never throttled)')
```

### Proposed

```python
# Modified (v13): Proposed method = hybrid candidate evaluation + Lyapunov relay
#   - movement: per drone the actor proposes K candidate moves; EACH candidate is
#     scored with the true system model (channel -> inner delivery subproblem ->
#     Lyapunov score) and the best is executed (works with zero training)
#   - delivery: per slot the rates D_i solve the inner subproblem (critical-price
#     bisection + vertex rounding) under the hard backhaul constraint tau*R_bh_u;
#     bandwidth allocation itself is NOT compressed
#   - learning: REINFORCE reinforces every sampled candidate with advantage
#     (J_cand - mean(J)) / std, i.e. self-critical style; reward log kept as before
from relay_hybrid_controller import HybridCandidateController

# Initiate the UAV Swarm (relay drones with the original 3D movement) 
uav_swarm_proposed = [UAV(name= f'UAV {i} (Proposed)', x_init=init_locations_uav[i][0], y_init=init_locations_uav[i][1], z_init=z_init_proposed, 
                        xlim = xlim_uav[i], ylim = ylim_uav[i], speedxy_max=uav_speedxy_max, speedz_max=uav_speedz_max) for i in range(0,n_uavs)]

# Hybrid Lyapunov relay controller (see relay_hybrid_controller.py)
ctrl = HybridCandidateController(
    n_users=n_users, n_uavs=n_uavs, boundary=boundary, grid_size=grid_size,
    queue_norm=queue_norm_Mb, V_lyap=lyapunov_param, tau=slot_len,
    ss_min_types=ss_min_types, ss_max_types=ss_max_types,
    mos_weights=mos_type_weights, type_rates=ON_data_arrival_mean_Mb, d0=d0,
    sigma_decay_slots=n_slots, seed=0,
    n_candidates=20, eval_sigma=1.1
)
ctrl.bind_models(cal_channel_fading_gain, cal_backhaul_capacity_Mb, pTx_downlink, noise_pw_total)

# Initiate users of the Proposed swarm with movements and traffic 
users_proposed = [User(f'User {i}') for i in range(n_users)]
load_user_properties(
    users_proposed,
    xaxis_all, yaxis_all,                               # location
    arrival_traffic_Mb, traffic_state, traffic_type,    # traffic
    ema_traffic_Mb, lambd_Mb_alluser
)

# Alias for ease of code maintenance 
uavs = uav_swarm_proposed
users = users_proposed          
    
# Simulation starts!  
bh_capa_log, bh_util_log, reward_log = [], [], []
bestJ_log = []
for t in range(0, n_slots):         # t = timeslot
    if t%100==0: 
        print(f"t={t}")
    # (0) Collect system statistics 
    queue_length_Mb = np.array([user.queue_length_Mb[t-1] for user in users]) if t > 0 else np.zeros(n_users)  # (n_users,)
    traffic_Mb = np.array([user.arrival_traffic_Mb[t] for user in users])       # shape=(n_users,)
    isActive_all = find_active_users(queue_len_Mb=queue_length_Mb, incoming_traffic_Mb=traffic_Mb)  # (n_users,)
    traffic_state_all = np.array([user.traffic_state[t] for user in users])     # shape=(n_users,)
    mos_all = np.array([user.mos[t-1] for user in users]) if t > 0 else 5 * np.ones(n_users)
    ld_Mbps = np.array([0 if (user.traffic_type[t] == -1) else ON_data_arrival_mean_Mb[user.traffic_type[t]] for user in users])
    
    # (1) User Clustering: based on locations of users 
    loc_users = np.hstack((
        np.expand_dims(xaxis_all[:,t], axis=1),
        np.expand_dims(yaxis_all[:,t], axis=1),
        np.zeros(shape=(n_users,1))
    ))    # shape = (n_users, 3)
    cmat, cids, c_centroids = kmeans_clustering(loc_users)        # cluster matrix, ids, and centroid
    
    # (2) Access bandwidth pools (same rule as the baseline methods: 30/60/10 by type)
    bw_alloc_Hz_all = np.zeros(shape=(n_users,))
    ttype_all_t = traffic_type[:, t]
    for k in range(len(bw_type_pool_Hz)): 
        ids_k = np.where(ttype_all_t == k)[0]
        active_k = ids_k[isActive_all[ids_k]]
        if len(active_k) > 0: 
            bw_alloc_Hz_all[active_k] = bw_type_pool_Hz[k] / len(active_k)
    bw_alloc_all = bw_alloc_Hz_all / bw_per_user      # fraction of the reference per-user bw
    
    # (3) UAV movement (hybrid): actor proposes K candidates per drone, each scored with
    # the true system model (channel -> D-subproblem -> Lyapunov score); best is executed
    heatmap_t = gen_heatmap(xaxis_all[:,t], yaxis_all[:,t], queue_length_Mb + traffic_Mb, queue_norm_Mb)
    uav_locs = np.array([[uav.x[t-1], uav.y[t-1], uav.z[t-1]] for uav in uavs]) if t>0 \
               else np.array([[uav.x[0], uav.y[0], uav.z[0]] for uav in uavs])
    ctrl.begin_slot(heatmap_t, uav_locs, np.array([airship_x, airship_y, airship_z]))
    for k, uav in enumerate(uavs):
        best_a, best_J = ctrl.decide_and_score(
            k=k, uav_loc=uav_locs[k], airship_loc=np.array([airship_x, airship_y, airship_z]),
            heatmap=heatmap_t, user_x=xaxis_all[:,t], user_y=yaxis_all[:,t],
            bw_alloc_Hz=bw_alloc_Hz_all, Q=queue_length_Mb, A=traffic_Mb, ld=ld_Mbps,
            types=traffic_type[:,t], cids=cids,
            speed_xy=uav_speedxy_max, speed_z=uav_speedz_max, zmin=z_min, zmax=z_max,
            boundary=boundary)
        bestJ_log.append(best_J)
        vx, vy, vz = best_a
        xt = uav.x[t-1] + vx*slot_len
        yt = uav.y[t-1] + vy*slot_len
        zt = uav.z[t-1] + vz*slot_len
        uav.x[t] = adjust_location(xt, bound=boundary)
        uav.y[t] = adjust_location(yt, bound=boundary)
        uav.z[t] = adjust_altitude(zt, zmin=z_min, zmax=z_max)
        uav.velocity[t] = np.sqrt(vx**2 + vy**2 + vz**2)
        uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=uav.velocity[t])[0]
        uav.vqlen_prop[t] = update_vqlen_prop(uav.vqlen_prop[t-1], uav.pw_prop_W[t]) if t > 0 else 0.0
    
    # (3.1) Periodic actor update (REINFORCE on the Lyapunov drift-minus-reward)
    if (t >= t_training_start) and (t % training_interval == 0): 
        ctrl.train_step()
    
    # (4) Access rates at the new positions (Mb per slot)
    R_access = np.zeros(shape=(n_users,))
    for k in range(n_uavs): 
        ids_k = np.where(cids == k)[0]
        if len(ids_k) == 0: 
            continue
        g_k = cal_channel_fading_gain(xaxis_all[ids_k, t], yaxis_all[ids_k, t], uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        R_access[ids_k] = bw_alloc_Hz_all[ids_k] * np.log2(1 + pTx_downlink * g_k / noise_pw_total) * slot_len / 1e6
    
    # (5) Backhaul rates and the inner delivery subproblem (bisection + vertex rounding)
    R_bh = np.array([cal_backhaul_capacity_Mb(uav.x[t], uav.y[t], uav.z[t]) for uav in uavs])
    D_opt, S_u = ctrl.solve_delivery(
        R_access=R_access, R_backhaul=R_bh, Q=queue_length_Mb, A=traffic_Mb,
        cids=cids, active=isActive_all, ld=ld_Mbps, types=traffic_type[:, t]
    )
    bh_capa_log.append(np.mean(R_bh))
    bh_util_log.append(np.sum(S_u) / max(np.sum(R_bh), 1e-9))
    
    # (6) User updates with the delivery rates D
    mos_post = np.full(shape=(n_users,), fill_value=np.nan)
    for id, user in enumerate(users):
        cid = cids[id]
        user.cluster[t] = cids[id]
        user.active_state[t] = isActive_all[id]
        user.alpha[t] = bw_alloc_all[id]
        user.chgains_downlink[t] = cal_channel_fading_gain(user.x[t], user.y[t], uavs[cid].x[t], uavs[cid].y[t], uavs[cid].z[t])
        user.channel_capacity_Mb[t] = R_access[id]
        qlen_prev_Mb = user.queue_length_Mb[t-1] if t > 0 else 0
        user.downlink_throughput_Mb[t] = D_opt[id]
        user.queue_length_Mb[t] = update_queue_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], D_opt[id])
        user.ema_qlen_Mb[t] = alpha_ema*user.queue_length_Mb[t] + (1-alpha_ema)*user.ema_qlen_Mb[t-1]
        ttype_t = int(user.traffic_type[t])
        ld_type_Mbps = ON_data_arrival_mean_Mb[ttype_t] if ttype_t >= 0 else user.lambd_Mb
        user.mos[t] = mos_func(qlen=user.ema_qlen_Mb[t], lambda_=ld_type_Mbps, ss_min_arr=ss_min_types[ttype_t], ss_max_arr=ss_max_types[ttype_t]) if (user.traffic_state[t]==True) else np.nan
        user.delay[t] = d0 + user.ema_qlen_Mb[t] / ld_type_Mbps
        # post-delivery MOS used inside the Lyapunov reward
        if user.traffic_state[t] == True: 
            qlen_post = np.max([qlen_prev_Mb + user.arrival_traffic_Mb[t] - D_opt[id], 0])
            mos_post[id] = mos_func(qlen=qlen_post, lambda_=ld_type_Mbps, ss_min_arr=ss_min_types[ttype_t], ss_max_arr=ss_max_types[ttype_t])
    
    # (7) Lyapunov reward, virtual-queue update, store the transition for REINFORCE
    reward_t = ctrl.compute_reward(Q=queue_length_Mb, A=traffic_Mb, D=D_opt, S_u=S_u,
                                   R_bh=R_bh, mos_mean=np.nanmean(mos_post))
    # commit the candidate records (state snapshot taken in begin_slot at decision time),
    # then update the backhaul virtual queues
    ctrl.commit_slot(reward_t)
    ctrl.update_virtual_queues(S_u=S_u, R_bh=R_bh)
    reward_log.append(reward_t)

print(f"Backhaul [Proposed-Lyapunov]: avg capacity/drone = {np.mean(bh_capa_log):.1f} Mb/slot, avg utilization = {np.mean(bh_util_log):.3f}")
print(f"Lyapunov reward: mean = {np.mean(reward_log):.2f}, last-100 avg = {np.mean(reward_log[-100:]):.2f}")
print(f"Backhaul virtual queue Z (Mb): final = {np.round(ctrl.Z, 2)}, max ever = {np.round(ctrl.Z_max, 2)}")
print(f"Candidate evaluation score J: mean = {np.mean(bestJ_log):.2f}, last-100 avg = {np.mean(bestJ_log[-100:]):.2f}")
```

```python
#
# SIMULATION
#


# Initiate the UAV Swarm 
uav_swarm_a2c = [
    UAV(
        name= f'UAV {i} (A2C)', x_init=init_locations_uav[i][0], y_init=init_locations_uav[i][1], z_init=z_init, 
        xlim = xlim_uav[i], ylim = ylim_uav[i], speedxy_max=uav_speedxy_max, speedz_max=uav_speedz_max
    ) for i in range(0,n_uavs)
]

# Initiate users of the Stationary UAV Swarm with movements and traffic 
users_a2c = [User(f'User {i}') for i in range(n_users)]
load_user_properties(
    users_a2c, xaxis_all, yaxis_all, arrival_traffic_Mb, traffic_state,
    traffic_type, ema_traffic_Mb, lambd_Mb_alluser
)

# Alias for ease of code maintenance 
uavs = uav_swarm_a2c
users = users_a2c
    
# Simulation starts!  
bh_capa_log, bh_scale_log = [], []
for t in range(0, n_slots):         # t = timeslot
    if t%100==0: 
        print(f"t={t}")
    # (0) Collect system statistics 
    queue_length_Mb = np.array([user.queue_length_Mb[t-1] for user in users]) if t>0 else np.zeros(n_users)     # shape=(n_users,)
    traffic_Mb = np.array([user.arrival_traffic_Mb[t] for user in users])       # shape=(n_users,)
    isActive_all = find_active_users(queue_len_Mb=queue_length_Mb, incoming_traffic_Mb=traffic_Mb)  # shape=(n_users,), =1 if qlen + traffic > 0
    mos_all = np.array([user.mos[t-1] for user in users]) if t > 0 else 5 * np.ones(n_users)

    # (1) User Clustering 
    loc_users = np.hstack((np.expand_dims(xaxis_all[:,t], axis=1), np.expand_dims(yaxis_all[:,t], axis=1), np.zeros(shape=(n_users,1))))    # shape = (n_users, 3)
    loc_uavs = np.array([[uav.x[t-1], uav.y[t-1], uav.z[t-1]] for uav in uavs]) if t>0 \
                else  np.array([[uav.x[0], uav.y[0], uav.z[0]] for uav in uavs])  # shape = (n_uavs, 3)
    cmat, cids, c_centroids = kmeans_clustering(loc_users, n_clusters=n_uavs)
    
    # (2) Bandwidth Allocation (modified): the 50 MHz pool is split across the 3 service
    # types with shares 30% (Web) / 60% (Video) / 10% (Messaging); in each slot a type's
    # pool is shared equally by its active users. alpha = user bw / bw_per_user (reference),
    # so channel capacity = alpha * bw_per_user * log2(1+SNR) = user_bw * log2(1+SNR).
    bw_alloc_Hz_all = np.zeros(shape=(n_users,))
    ttype_all_t = traffic_type[:, t]
    for k in range(len(bw_type_pool_Hz)): 
        ids_k = np.where(ttype_all_t == k)[0]
        active_k = ids_k[isActive_all[ids_k]]
        if len(active_k) > 0: 
            bw_alloc_Hz_all[active_k] = bw_type_pool_Hz[k] / len(active_k)
    bw_alloc_all = bw_alloc_Hz_all / bw_per_user      # fraction of the reference per-user bw
    
    # (3) UAV Updates: Movement and Energy Consumption  
    if t > 0 and (not uav_fixed_placement): 
        for cid, uav in enumerate(uavs):        # cid = cluster id 
            if t < t_training_start: 
                vx, vy = 0, 0  # no movement 
            else:
                xlocs = xaxis_all[cids==cid, t]
                ylocs = yaxis_all[cids==cid, t]
                observation = get_obs_flattened(
                    locs={
                        'self': np.asarray([uav.x[t-1], uav.y[t-1]]).reshape(2, 1),
                        'user': np.asarray([xlocs, ylocs])
                    },
                    bound=boundary,
                    grid_size=grid_size,
                    grid_norm=user_counter_norm,
                    mode="sarl"
                )
                action, _states = model.predict(observation, deterministic=deterministic)
                vx = action[0] * uav_vxy_a2c
                vy = action[1] * uav_vxy_a2c

            uav.x[t] = uav.x[t-1] + vx * slot_len
            uav.y[t] = uav.y[t-1] + vy * slot_len
            uav.z[t] = uav.z[0]
            uav.velocity[t] = np.sqrt(vx**2 + vy**2)
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=uav.velocity[t])[0]


    # Modified: aerial BSs are pinned at the center of the area (altitude fixed at 300 m)
    if uav_fixed_placement: 
        for uav in uavs: 
            uav.x[t], uav.y[t], uav.z[t] = uav.x[0], uav.y[0], uav.z[0]
            uav.velocity[t] = 0
            uav.pw_prop_W[t] = cal_uav_propulsion_energy(V=0)[0]
    # (3.5) Modified: backhaul constraint - airship -> drone k hard-caps the drone's
    # total served downlink rate (excess stays queued; bandwidth allocation untouched).
    bh_scale = np.ones(shape=(n_uavs,))
    bh_capa_k = np.zeros(shape=(n_uavs,))
    for k in range(n_uavs): 
        ids_k = np.where(cids == k)[0]
        if len(ids_k) == 0: 
            continue
        g_k = cal_channel_fading_gain(xaxis_all[ids_k, t], yaxis_all[ids_k, t], uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        capa_k = bw_alloc_Hz_all[ids_k] * np.log2(1 + pTx_downlink * g_k / noise_pw_total) * slot_len / 1e6   # Mb/slot
        offered_k = np.sum(np.minimum(capa_k, queue_length_Mb[ids_k] + traffic_Mb[ids_k]))
        bh_capa_k[k] = cal_backhaul_capacity_Mb(uavs[k].x[t], uavs[k].y[t], uavs[k].z[t])
        if offered_k > bh_capa_k[k]: 
            bh_scale[k] = bh_capa_k[k] / offered_k
    bh_capa_log.append(np.mean(bh_capa_k))
    bh_scale_log.append(np.min(bh_scale))
    # (4) User Updates  
    for id, user in enumerate(users): 
        user.cluster[t] = cids[id]
        cid = cids[id]      # id of the user's cluster  
        user.active_state[t] = isActive_all[id]
        user.alpha[t] = bw_alloc_all[id]       # fixed bandwidth for each user 
        user.chgains_downlink[t] = cal_channel_fading_gain(user.x[t], user.y[t], uavs[cid].x[t], uavs[cid].y[t], uavs[cid].z[t])
        user.channel_capacity_Mb[t] = cal_channel_capacity_Mb(user.chgains_downlink[t], user.alpha[t])[0]
        
        qlen_prev_Mb = user.queue_length_Mb[t-1] if t > 0 else 0
        depart_Mb = cal_downlink_throughput_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], user.channel_capacity_Mb[t]) * bh_scale[cid]
        user.queue_length_Mb[t] = update_queue_Mb(qlen_prev_Mb, user.arrival_traffic_Mb[t], depart_Mb)
        user.downlink_throughput_Mb[t] = depart_Mb
        user.ema_qlen_Mb[t] = alpha_ema*user.queue_length_Mb[t] + (1-alpha_ema)*user.ema_qlen_Mb[t-1]
        ttype_t = int(user.traffic_type[t])   # service type active in this slot (-1 if OFF)
        ld_type_Mbps = ON_data_arrival_mean_Mb[ttype_t] if ttype_t >= 0 else user.lambd_Mb
        user.mos[t] = mos_func(qlen=user.ema_qlen_Mb[t], lambda_=ld_type_Mbps, ss_min_arr=ss_min_types[ttype_t], ss_max_arr=ss_max_types[ttype_t]) if (user.traffic_state[t]==True) else np.nan
        user.delay[t] = d0 + user.ema_qlen_Mb[t] / ld_type_Mbps

print('Backhaul [A2C]: avg capacity/drone = ' +
      f'{np.mean(bh_capa_log):.1f} Mb/slot, time-avg of min bh_scale = {np.mean(bh_scale_log):.3f} (1.0 = never throttled)')
```

### KPI data

```python
def export_kpi_over_time(users, sim_folder_path, method_name):
    active_state_all = np.array([user.active_state for user in users])          # shape=(n_users,n_slots)
    traffic_state_all = np.array([user.traffic_state for user in users])
    traffic_Mb_all = np.array([user.arrival_traffic_Mb for user in users])      # shape=(n_users,n_slots)
    qlen_Mb_all = np.array([user.queue_length_Mb for user in users])            # shape=(n_users,n_slots)
    mos_all = np.array([user.mos for user in users])                            # shape=(n_users,n_slots)
    traffic_type_all = np.array([user.traffic_type for user in users])          # shape=(n_users,n_slots)
    chcapa_Mb_all = np.array([user.channel_capacity_Mb for user in users]) # shape=(n_users,n_slots)
     
    active_cnt = np.sum(active_state_all, axis=0)                       # shape=(n_users,)
    traffic_cnt = np.sum(traffic_state_all, axis=0)                     # shape=(n_users,)
    traffic_Mb_avg = np.sum(traffic_Mb_all, axis=0)/traffic_cnt         # shape=(n_users,)
    qlen_Mb_avg = np.sum(qlen_Mb_all, axis=0)/active_cnt                # shape=(n_users,)
    qlen_Mb_max = np.max(qlen_Mb_all,axis=0)                            # shape=(n_users,)
    mos_avg = np.nansum(mos_all, axis=0)/traffic_cnt                     # shape=(n_users,)
    # Modified: weighted-average MOS over the 3 service types (0.35 Web / 0.25 Video / 0.4
    # Messaging) replaces the previous sum over users. In each slot only types having
    # active users are counted, with the weights re-normalized over those types.
    n_types = len(mos_type_weights)
    mos_weighted = np.full(shape=(mos_all.shape[1],), fill_value=np.nan)
    mos_per_type = {tname: np.full(shape=(mos_all.shape[1],), fill_value=np.nan) for tname in traffic_type_names}
    for tt in range(mos_all.shape[1]): 
        w_sum, val_sum = 0.0, 0.0
        for k in range(n_types): 
            ids = np.where(traffic_type_all[:, tt] == k)[0]
            if len(ids) > 0 and np.any(~np.isnan(mos_all[ids, tt])): 
                mos_k = np.nanmean(mos_all[ids, tt])
                mos_per_type[traffic_type_names[k]][tt] = mos_k
                val_sum += mos_type_weights[k] * mos_k
                w_sum += mos_type_weights[k]
        if w_sum > 0: 
            mos_weighted[tt] = val_sum / w_sum
    mos_percentile25 = np.nanpercentile(mos_all, q=25, axis=0)          # shape=(n_users,)
    chcapa_Mb_avg = np.mean(chcapa_Mb_all, axis=0)     # shape=(n_users,)
    mos_avg_alltime = np.mean(mos_avg[t_training_start:])
    qlen_Mb_avg_alltime = np.mean(qlen_Mb_avg[t_training_start:])
    
    kpis = {
        'active-count':active_cnt,     
        'traffic-incoming-cnt':traffic_cnt,            
        'traffic-Mb-avg (incoming)':traffic_Mb_avg,
        'qlen-Mb-avg (active)':qlen_Mb_avg,                      # average of active users 
        'qlen-Mb-max':qlen_Mb_max, 
        'mos-avg (active)':mos_avg,                              # average of active users 
        'mos-weighted-avg':mos_weighted,
        'mos_percentile25':mos_percentile25,
        **{f'mos-avg-{tname}': mos_per_type[tname] for tname in traffic_type_names},
        'chcapa-Mb-avg (active)':chcapa_Mb_avg,                  # average of active users 
        'mos-avg-alltime':mos_avg_alltime, 
        'qlen-Mb-avg-alltime':qlen_Mb_avg_alltime 
    }
    
    df = pd.DataFrame(kpis)
    fpath = os.path.join(sim_folder_path, f'{method_name}-kpis-over-time.xlsx')
    df.to_excel(fpath)  
    print(f'Exported {method_name} to {fpath}')  
```
